```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("HurricaneMarketPrediction")
    .master("local[*]")
    .getOrCreate()
)
```

```python

sectors = [
    "energy",
    "financials",
    "industrials",
    "utilities"
]

datasets = {}

for sector in sectors:
    path = (
        f"../data/sector_datasets/"
        f"{sector}_hurricane_dataset.csv"
    )

    datasets[sector] = (
        spark.read
        .option("header", True)
        .option("inferSchema", True)
        .csv(path)
    )

    print(f"{sector}: {datasets[sector].count()} rows")
```
